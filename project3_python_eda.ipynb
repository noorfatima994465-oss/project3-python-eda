```python
# Project 3 — Student Performance EDA
## Introduction
"""This project analyzes student performance using Python and Pandas.
The dataset contains information about students such as age, gender, city,
study hours, attendance, assignment scores, exam scores, and final marks.
The main goal of this project is to clean the data, explore the dataset,
identify patterns, and find meaningful insights about student performance."""
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("students_eda_project3.csv")
#df.head()
#df.tail()
#df.shape
#df.columns
#df.info()
#df.describe()
#df.isnull().sum()
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
df["Study_Hours"] = df["Study_Hours"].fillna(df["Study_Hours"].mean())
df["City"] = df["City"].fillna(df["City"].mode()[0])
df.isnull().sum()
df["City"]
## What are the highest marks?
df.sort_values("Final_Marks", ascending=False).head()
## Which student study the most ?
df.sort_values("Study_Hours", ascending=False).head()
## What is the avg final marks?
df["Final_Marks"].mean()
## avg marks by gender?
df.groupby("Gender")["Final_Marks"].mean()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("students_eda_project3.csv")
## graphs
plt.scatter(df["Study_Hours"], df["Final_Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Final Marks")
plt.title("Study Hours vs Final Marks")
plt.show()
###
df.groupby("Gender")["Final_Marks"].mean().plot(kind="bar")
plt.title("Average Final Marks by Gender")
plt.ylabel("Average Marks")
plt.show()

```

```python
## Insights

"""1. Students who study more hours generally tend to have higher final marks.

   2. The average final marks differ between cities.

   3. Attendance appears to have a relationship with student performance.

   4. Students with higher exam scores generally have higher final marks."""
```
